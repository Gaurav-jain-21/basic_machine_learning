##KKN imputation

```python
from sklearn.impute import KNNImputer
```

```python
data=pd.DataFrame(
    {
        'Age':[25,30,np.nan, 40,35,np.nan, 28],
        'Salary':[50000,60000,55000,np.nan, 65000,52000,np.nan],
        'Experience':[2,5,3,10,7,4,3]
    }
)
```

```python
data
```

```python
imputer= KNNImputer(n_neighbors=5)
data_imputed= imputer.fit_transform(data)
print(data_imputed)
df_imputed = pd.DataFrame(data_imputed, columns=data.columns)

print(df_imputed)
```

```python
data={
    'Customer': ['Riya','Meena','Tajes','Divya','Paras'],
    "Satisfaction":["Low",'High','Median','Very High','Low']
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
#Label Encoding
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
rank=[['Low','Median','High','Very High']]
```

```python
label=LabelEncoder()
ordinal= OrdinalEncoder(categories=rank)
df['Ordinal_Encoding'] = ordinal.fit_transform(df[['Satisfaction']])
df["Label_Encodeing"]= label.fit_transform(df[['Satisfaction']])
print(df)
```

```python
data={
    'Salary': [30000,50000,70000,200000],
    'Experience': [2,5,7,10],
    'Age':[25,30,35,60]
}
df=pd.DataFrame(data)
print(df)
```

```python
#find the minimy and maximum of all the features
print(min(df['Salary']))
print(max(df['Salary']))
print(min(df['Experience']))
print(max(df['Experience']))
print(min(df['Age']))
print(max(df['Age']))
```

```python
from sklearn.preprocessing import MaxAbsScaler
```

```python
maxabs_scale=MaxAbsScaler()
df_maxabs= pd.DataFrame(maxabs_scale.fit_transform(df), columns=df.columns)
print(df_maxabs)
```

```python
from sklearn.preprocessing import MinMaxScaler
minabs_scale= MinMaxScaler()
df_minabs= pd.DataFrame(minabs_scale.fit_transform(df), columns=df.columns)
print(df_minabs)
```

```python
from sklearn.preprocessing import Normalizer
l2_normalizer= Normalizer(norm='l2')
df_l2= pd.DataFrame(l2_normalizer.fit_transform(df), columns=df.columns)
print(df_l2)
```

```python
from sklearn.preprocessing import StandardScaler
z_score= StandardScaler()
df_z= pd.DataFrame(z_score.fit_transform(df), columns=df.columns)
print(df_z)
```
