```python
import pandas as pd
import numpy as np
```

```python
data= np.array([[2,3,4],[3,5,6],[5,8,9],[6,9,11],[7,10,13]])
```

```python
data
```

```python
mean_X= np.mean(data,axis=0)
print(mean_X)
```

```python
X_centered = data-mean_X
```

```python
X_centered
```

```python
cov_matrix= np.cov(X_centered, rowvar=False)
print(cov_matrix)
```

```python
eigenvalue, eigenvectors= np.linalg.eig(cov_matrix)
print(eigenvalue)
print(eigenvectors)
```

```python
sorted_indices = np.argsort(eigenvalue)[::-1]
eigenvalue= eigenvalue[sorted_indices]
eigenvectors= eigenvectors[:,sorted_indices]
```

```python
X_transformed_1= X_centered @ eigenvectors[:,0]
```

```python
X_transformed_2= X_centered @ eigenvectors[:,:2]
```

```python
print("transformed Data (3 to 1)")
print(X_transformed_1)
print("\n Tansformed Data (3 to 2)")
print(X_transformed_2)
```
